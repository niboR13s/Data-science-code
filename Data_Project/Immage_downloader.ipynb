````python
import os
import shutil
from icrawler.builtin import BingImageCrawler

# Check imports
print("icrawler loaded successfully.")
````

## 1. Configuration

Configure your search terms here. 
* **Note:** `icrawler` is very fast. 
* **Output:** It will automatically create folders for each search term inside the `MAIN_OUTPUT_DIR`.

````python
# --- CONFIGURATION ---

SEARCH_QUERIES = [
    # "car with open door",
    # "car trunk open",
    # "car hood open",
    # "car with open door top view",
    # "car trunk open top view",
    # "car hood open top view"
    # "parked car side view",
    # "car side profile",
    # "car front view",
    # "parked car front",
    # "car rear view",
    # "cars parked on street",
    # "car dealership lot"
    "car breakdown highway",
    "checking car oil",
    "car bonnet open",
    "cat bonnet ajar",
    "car boot ajar",
    "loading groceries into car",
    "putting luggage in car trunk",
    "car boot open",
    "unloading shopping bags from car",
    "person getting out of car",
    "getting into car",
    "car door ajar",
    "taxi passenger entering",
    "car door open parking lot",
    "high angle view car open door",


]

# How many images per query?
IMAGES_PER_QUERY = 300

# Main folder for dataset
MAIN_OUTPUT_DIR = "dataset_raw"

print(f"Configuration Ready. Target: {MAIN_OUTPUT_DIR}")
````

## 2. The Crawler Engine

This block iterates through your queries and starts a `GoogleImageCrawler` for each one.

**Key settings:**
* `feeder_threads`: Number of threads to parse URLS.
* `parser_threads`: Number of threads to download images.
* `storage`: Where to save the files.

````python
def run_crawler(queries, limit, root_dir):
    print("--- STARTING CRAWLER ---")
    
    for query in queries:
        print(f"\nProcessing: '{query}'...")
        
        # Define the subfolder for this specific query
        # We replace spaces with underscores for cleaner folder names
        subfolder = query.replace(" ", "_")
        save_path = os.path.join(root_dir, subfolder)
        
        # Initialize the crawler
        google_crawler = BingImageCrawler(
            feeder_threads=1,
            parser_threads=2,
            downloader_threads=4,
            storage={'root_dir': save_path}
        )
        
        # Start the crawl
        # keyword: the search term
        # max_num: max images to download
        google_crawler.crawl(keyword=query, max_num=limit)
        
    print("\n--- ALL DOWNLOADS COMPLETE ---")

# Execute
run_crawler(SEARCH_QUERIES, IMAGES_PER_QUERY, MAIN_OUTPUT_DIR)
````

## 3. Post-Processing & Cleanup

Since scrapers can sometimes download corrupt files, we perform a cleanup pass to remove empty files or files that are not valid images.

````python
def cleanup_dataset(root_dir):
    print("--- CLEANING DATASET ---")
    deleted_count = 0
    renamed_count = 0
    
    if not os.path.exists(root_dir):
        print("Directory not found.")
        return

    for folder in os.listdir(root_dir):
        folder_path = os.path.join(root_dir, folder)
        if not os.path.isdir(folder_path):
            continue
            
        files = os.listdir(folder_path)
        print(f"Checking {folder}... ({len(files)} files)")
        
        for idx, filename in enumerate(files):
            filepath = os.path.join(folder_path, filename)
            
            # 1. Check size (Delete < 5KB)
            if os.path.getsize(filepath) < 5120:
                os.remove(filepath)
                deleted_count += 1
                continue
            
            # 2. Rename for consistency (class_001.jpg)
            # icrawler saves as 000001.jpg, we want a class prefix
            ext = os.path.splitext(filename)[1]
            new_name = f"{folder}_{idx:03d}{ext}"
            new_path = os.path.join(folder_path, new_name)
            
            # Only rename if not already renamed
            if filename != new_name:
                try:
                    os.rename(filepath, new_path)
                    renamed_count += 1
                except:
                    pass

    print(f"\nCleanup Done. Deleted {deleted_count} junk files.")

# Run cleanup
cleanup_dataset(MAIN_OUTPUT_DIR)
````
