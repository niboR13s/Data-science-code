# A very simple Neural Network
In this notebook we will do some test with a very simple network. It mimics the XOR function.

```python
import tensorflow as tf
import numpy as np
import random
import matplotlib.pyplot as plt
```

```python
data_configs = {
    "XOR": {
        "X": np.array([[0,0], [0,1], [1,0], [1,1]]),
        "y": np.array([[0],   [1],   [1],   [0]])
    },
    "AND": {
        "X": np.array([[0,0], [0,1], [1,0], [1,1]]),
        "y": np.array([[0],   [0],   [0],   [1]])
    },
    "OR": {
        "X": np.array([[0,0], [0,1], [1,0], [1,1]]),
        "y": np.array([[0],   [1],   [1],   [1]])
    }
}
```

## making the models

```python

models = {
"model_XOR_0" : tf.keras.Sequential([
    tf.keras.layers.Dense(2, input_dim=2, activation='sigmoid'),  # hidden layer
    tf.keras.layers.Dense(1, activation=None)                    # output layer without activation
]),

"model_XOR_1" : tf.keras.Sequential([
    tf.keras.layers.Dense(2, input_dim=2, activation='sigmoid'),  # hidden layer
    tf.keras.layers.Dense(1, activation=None)                    # output layer without activation
]),

"model_XOR_2" : tf.keras.Sequential([
    tf.keras.layers.Dense(2, input_dim=2, activation='sigmoid'),  # hidden layer
    tf.keras.layers.Dense(1, activation=None)                    # output layer without activation
]),

"model_AND_0" : tf.keras.Sequential([
    tf.keras.layers.Dense(2, input_dim=2, activation='sigmoid'),  # hidden layer
    tf.keras.layers.Dense(1, activation=None)                    # output layer without activation
]),

"model_OR_0" : tf.keras.Sequential([
    tf.keras.layers.Dense(2, input_dim=2, activation='sigmoid'),  # hidden layer
    tf.keras.layers.Dense(1, activation=None)                    # output layer without activation
]),
}
```

```python
for model_name, model in models.items():
    print(f"\n{'='*60}")
    print(f"🚀 Training: {model_name}")
    print(f"{'='*60}")

    parts = model_name.split('_')
    logic_type = parts[1]
    
    print(f"   -> Detected Logic Type: {logic_type}")

    if logic_type in data_configs:
        X = data_configs[logic_type]["X"]
        y = data_configs[logic_type]["y"]
    else:
        print(f"⚠️ Warning: Unknown logic type '{logic_type}'. Skipping...")
        continue

    # --- COMPILE ---
    model.compile(optimizer=tf.keras.optimizers.Adam(learning_rate=0.1),
                  loss='mse',
                  metrics=['mse', 'accuracy']) 

    # --- TRAIN ---
    history = model.fit(X, y, epochs=400, verbose=0)
    

    final_loss = history.history['loss'][-1]
    print(f"   -> Final Loss: {final_loss:.4f}")

    # --- PLOT LEARNING CURVE ---
    plt.figure(figsize=(6, 4))
    plt.plot(history.history['loss'], label='Loss (MSE)')
    plt.title(f'Learning Curve: {model_name}')
    plt.xlabel('Epoch')
    plt.ylabel('Loss')
    plt.grid(True, linestyle='--', alpha=0.5)
    plt.legend()
    plt.show()

    # --- TEST PREDICTIONS ---
    predictions = model.predict(X, verbose=0)
    print("   -> Predictions (raw):")
    # Even netjes printen naast de input
    for i in range(len(X)):
        print(f"      Input {X[i]} -> Pred: {predictions[i][0]:.4f} (Target: {y[i][0]})")

    # --- PRINT WEIGHTS ---
    print("\n   -> Model Weights:")
    for i, layer in enumerate(model.layers):
        weights, biases = layer.get_weights()
        print(f"      Layer {i+1} Weights:\n{weights}")
        print(f"      Layer {i+1} Biases: \n{biases}\n")
```
